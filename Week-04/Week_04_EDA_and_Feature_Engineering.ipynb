```python

# WEEK 04: EDA AND FEATURE ENGINEERING
# Beauty & Wellness - Yuva Data Science Internship

import os
import zipfile
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import spearmanr, mannwhitneyu, chi2_contingency
from IPython.display import display

warnings.filterwarnings("ignore")

# ============================================================
# 1. FILE PATHS
# ============================================================

RAW_PATH = "/content/drive/MyDrive/Colab Notebooks/Dataset/reviews_sample.parquet"
CLEAN_PATH = "/content/drive/MyDrive/Colab Notebooks/Dataset/beauty_reviews_cleaned.csv"
SUMMARY_PATH = "/content/drive/MyDrive/Colab Notebooks/Dataset/week3_cleaning_summary.csv"

OUTPUT_DIR = "/content/drive/MyDrive/Colab Notebooks/Dataset/Week-04-Outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ============================================================
# 2. LOAD DATA
# ============================================================

raw_df = pd.read_parquet(RAW_PATH)
clean_df = pd.read_csv(CLEAN_PATH)

print("=" * 65)
print("WEEK 04: EDA AND FEATURE ENGINEERING")
print("=" * 65)

print("\nRAW DATA SHAPE:", raw_df.shape)
print("CLEANED DATA SHAPE:", clean_df.shape)

print("\nRAW DATA COLUMNS:")
print(raw_df.columns.tolist())

print("\nCLEANED DATA COLUMNS:")
print(clean_df.columns.tolist())

print("\nCLEANED DATA SAMPLE:")
display(clean_df.head())

# ============================================================
# 3. STANDARDIZE COLUMN NAMES
# ============================================================

raw_df.columns = raw_df.columns.astype(str).str.strip().str.lower()
clean_df.columns = clean_df.columns.astype(str).str.strip().str.lower()

column_aliases = {
    "userid": "user_id",
    "itemid": "item_id",
    "product_id": "item_id",
    "asin": "item_id",
    "parent_asin": "item_id",
    "review_title": "title",
    "review_text": "text",
    "helpful_votes": "helpful_vote",
    "verified": "verified_purchase"
}

raw_df.rename(columns=column_aliases, inplace=True)
clean_df.rename(columns=column_aliases, inplace=True)

# Remove duplicate column names, if any
raw_df = raw_df.loc[:, ~raw_df.columns.duplicated()].copy()
clean_df = clean_df.loc[:, ~clean_df.columns.duplicated()].copy()

# ============================================================
# 4. CONVERT DATA TYPES SAFELY
# ============================================================

for df in [raw_df, clean_df]:
    if "rating" in df.columns:
        df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

    if "helpful_vote" in df.columns:
        df["helpful_vote"] = pd.to_numeric(
            df["helpful_vote"], errors="coerce"
        )

# Unix timestamps are stored in seconds in the raw dataset
if "timestamp" in raw_df.columns:
    raw_df["timestamp"] = pd.to_datetime(
        pd.to_numeric(raw_df["timestamp"], errors="coerce"),
        unit="s",
        errors="coerce",
        utc=True
    )

# Convert Unix timestamps in cleaned data
if "timestamp" in clean_df.columns:
    clean_df["timestamp"] = pd.to_datetime(
        pd.to_numeric(clean_df["timestamp"], errors="coerce"),
        unit="s",
        errors="coerce",
        utc=True
    )

# Convert the separate readable date column, if present
if "review_date" in clean_df.columns:
    clean_df["review_date"] = pd.to_datetime(
        clean_df["review_date"], errors="coerce", utc=True
    )

print("\nDATE CONVERSION CHECK")
date_columns = [
    c for c in ["timestamp", "review_date"] if c in clean_df.columns
]
if date_columns:
    display(clean_df[date_columns].head())

# ============================================================
# 5. VALIDATE REQUIRED COLUMNS
# ============================================================

if "rating" not in clean_df.columns:
    raise ValueError(
        "Cleaned dataset does not contain a rating column."
    )

if "rating" not in raw_df.columns:
    print("WARNING: Raw dataset has no rating column.")

# ============================================================
# 6. DATA QUALITY COMPARISON
# ============================================================

quality_rows = []

for name, df in [("Raw", raw_df), ("Cleaned", clean_df)]:
    quality_rows.append({
        "dataset": name,
        "rows": len(df),
        "columns": len(df.columns),
        "missing_cells": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum()),
        "missing_ratings": (
            int(df["rating"].isna().sum())
            if "rating" in df.columns else None
        ),
        "invalid_ratings": (
            int((~df["rating"].between(1, 5)).sum())
            if "rating" in df.columns else None
        )
    })

quality_report = pd.DataFrame(quality_rows)

print("\nDATA QUALITY REPORT")
display(quality_report)

quality_report.to_csv(
    os.path.join(OUTPUT_DIR, "data_quality_comparison.csv"),
    index=False
)

print("\nROW COUNT CHANGE:", len(raw_df) - len(clean_df))

# ============================================================
# 7. DESCRIPTIVE STATISTICS
# ============================================================

print("\nCLEANED DATA DESCRIPTIVE STATISTICS")
display(clean_df.describe(include="all").T)

print("\nMISSING VALUES BY COLUMN")
missing_report = clean_df.isna().sum().sort_values(ascending=False)
display(missing_report.to_frame("missing_count"))

missing_report.to_csv(
    os.path.join(OUTPUT_DIR, "missing_values_report.csv")
)

print("\nRATING DISTRIBUTION")
rating_counts = (
    clean_df["rating"].value_counts(dropna=False).sort_index()
)
display(rating_counts.to_frame("review_count"))

rating_counts.to_csv(
    os.path.join(OUTPUT_DIR, "rating_distribution.csv")
)

print("\nAVERAGE RATING:", round(clean_df["rating"].mean(), 3))
print("MEDIAN RATING:", round(clean_df["rating"].median(), 3))
print("RATING STANDARD DEVIATION:", round(clean_df["rating"].std(), 3))

# ============================================================
# 8. FEATURE ENGINEERING
# ============================================================

eda_df = clean_df.copy()

# Review text length features
text_column = "text_clean" if "text_clean" in eda_df.columns else (
    "text" if "text" in eda_df.columns else None
)

if text_column:
    review_text = eda_df[text_column].fillna("").astype(str)
    eda_df["review_length_chars"] = review_text.str.len()
    eda_df["review_length_words"] = review_text.str.split().str.len()
else:
    eda_df["review_length_chars"] = np.nan
    eda_df["review_length_words"] = np.nan

# Title length
title_column = "title_clean" if "title_clean" in eda_df.columns else (
    "title" if "title" in eda_df.columns else None
)

if title_column:
    eda_df["title_length_chars"] = (
        eda_df[title_column].fillna("").astype(str).str.len()
    )

# Rating-derived indicators
eda_df["positive_review"] = (eda_df["rating"] >= 4).astype("Int64")
eda_df["low_rating"] = (eda_df["rating"] <= 2).astype("Int64")
eda_df["neutral_rating"] = (eda_df["rating"] == 3).astype("Int64")

# Helpful votes
if "helpful_vote" in eda_df.columns:
    eda_df["has_helpful_vote"] = (
        eda_df["helpful_vote"].fillna(0) > 0
    ).astype(int)

# Verified purchase normalization
if "verified_purchase" in eda_df.columns:
    verified_map = {
        "true": 1, "1": 1, "yes": 1,
        "false": 0, "0": 0, "no": 0
    }

    eda_df["verified_purchase_clean"] = (
        eda_df["verified_purchase"]
        .astype(str)
        .str.strip()
        .str.lower()
        .map(verified_map)
    )

# Use timestamp first, then review_date as a fallback
if "timestamp" in eda_df.columns:
    date_series = pd.to_datetime(
        eda_df["timestamp"], errors="coerce", utc=True
    )
else:
    date_series = pd.Series(
        pd.NaT, index=eda_df.index, dtype="datetime64[ns, UTC]"
    )

if "review_date" in eda_df.columns:
    fallback_dates = pd.to_datetime(
        eda_df["review_date"], errors="coerce", utc=True
    )
    date_series = date_series.fillna(fallback_dates)

eda_df["analysis_date"] = date_series
eda_df["review_year"] = eda_df["analysis_date"].dt.year
eda_df["review_month"] = eda_df["analysis_date"].dt.month
eda_df["review_quarter"] = eda_df["analysis_date"].dt.quarter
eda_df["review_day_of_week"] = eda_df["analysis_date"].dt.dayofweek

print("\nENGINEERED FEATURES")
new_features = sorted(set(eda_df.columns) - set(clean_df.columns))
print(new_features)

# ============================================================
# 9. VISUALIZATION: RATING DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))
eda_df["rating"].dropna().value_counts().sort_index().plot(
    kind="bar", edgecolor="black"
)
plt.title("Distribution of Beauty Product Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.savefig(
    os.path.join(OUTPUT_DIR, "01_rating_distribution.png"), dpi=150
)
plt.show()

# ============================================================
# 10. VISUALIZATION: REVIEW LENGTH DISTRIBUTION
# ============================================================

lengths = eda_df["review_length_words"].dropna()

if not lengths.empty:
    upper_limit = lengths.quantile(0.99)
    lengths_plot = lengths[lengths <= upper_limit]

    plt.figure(figsize=(8, 5))
    plt.hist(lengths_plot, bins=40, edgecolor="black")
    plt.title("Distribution of Review Lengths")
    plt.xlabel("Words per Review")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.savefig(
        os.path.join(OUTPUT_DIR, "02_review_length_distribution.png"),
        dpi=150
    )
    plt.show()

# ============================================================
# 11. VISUALIZATION: REVIEW LENGTH BY RATING
# ============================================================

length_rating = eda_df[
    ["rating", "review_length_words"]
].dropna()

if not length_rating.empty:
    upper_limit = length_rating["review_length_words"].quantile(0.99)
    length_rating = length_rating[
        length_rating["review_length_words"] <= upper_limit
    ]

    if not length_rating.empty:
        plt.figure(figsize=(8, 5))
        length_rating.boxplot(
            column="review_length_words", by="rating"
        )
        plt.title("Review Length by Rating")
        plt.suptitle("")
        plt.xlabel("Rating")
        plt.ylabel("Review Length (Words)")
        plt.tight_layout()
        plt.savefig(
            os.path.join(OUTPUT_DIR, "03_review_length_by_rating.png"),
            dpi=150
        )
        plt.show()

# ============================================================
# 12. VERIFIED PURCHASE ANALYSIS
# ============================================================

if "verified_purchase_clean" in eda_df.columns:
    verified_data = eda_df.dropna(
        subset=["verified_purchase_clean", "rating"]
    )

    if not verified_data.empty:
        verified_summary = verified_data.groupby(
            "verified_purchase_clean"
        )["rating"].agg(["count", "mean", "median", "std"]).reset_index()

        print("\nRATING BY VERIFIED PURCHASE STATUS")
        display(verified_summary)

        verified_summary.to_csv(
            os.path.join(OUTPUT_DIR, "verified_purchase_summary.csv"),
            index=False
        )

        plt.figure(figsize=(8, 5))
        verified_data.boxplot(
            column="rating", by="verified_purchase_clean"
        )
        plt.title("Ratings by Verified Purchase Status")
        plt.suptitle("")
        plt.xlabel("Verified Purchase (0 = No, 1 = Yes)")
        plt.ylabel("Rating")
        plt.tight_layout()
        plt.savefig(
            os.path.join(OUTPUT_DIR, "04_verified_purchase_ratings.png"),
            dpi=150
        )
        plt.show()

# ============================================================
# 13. MONTHLY REVIEW TRENDS
# ============================================================

time_df = eda_df.dropna(subset=["analysis_date"]).copy()

if not time_df.empty:
    time_df["review_period"] = (
        time_df["analysis_date"].dt.to_period("M").astype(str)
    )

    monthly_summary = time_df.groupby("review_period").agg(
        review_count=("rating", "count"),
        average_rating=("rating", "mean"),
        low_rating_share=("low_rating", "mean")
    ).reset_index()

    print("\nMONTHLY REVIEW TRENDS")
    display(monthly_summary.tail(20))

    monthly_summary.to_csv(
        os.path.join(OUTPUT_DIR, "monthly_review_trends.csv"),
        index=False
    )

    plt.figure(figsize=(12, 5))
    plt.plot(
        monthly_summary["review_period"],
        monthly_summary["review_count"],
        marker="o"
    )
    plt.title("Monthly Review Volume")
    plt.xlabel("Month")
    plt.ylabel("Number of Reviews")
    plt.xticks(rotation=75)
    plt.tight_layout()
    plt.savefig(
        os.path.join(OUTPUT_DIR, "05_monthly_review_volume.png"),
        dpi=150
    )
    plt.show()

    plt.figure(figsize=(12, 5))
    plt.plot(
        monthly_summary["review_period"],
        monthly_summary["average_rating"],
        marker="o"
    )
    plt.title("Average Rating Over Time")
    plt.xlabel("Month")
    plt.ylabel("Average Rating")
    plt.xticks(rotation=75)
    plt.tight_layout()
    plt.savefig(
        os.path.join(OUTPUT_DIR, "06_monthly_average_rating.png"),
        dpi=150
    )
    plt.show()

# ============================================================
# 14. PRODUCT-LEVEL ANALYSIS
# ============================================================

product_column = next(
    (
        col for col in ["item_id", "product_id", "asin", "parent_asin"]
        if col in eda_df.columns
    ),
    None
)

if product_column:
    product_summary = eda_df.groupby(product_column).agg(
        review_count=("rating", "count"),
        average_rating=("rating", "mean"),
        rating_std=("rating", "std"),
        low_rating_share=("low_rating", "mean")
    ).reset_index()

    product_summary = product_summary.sort_values(
        "review_count", ascending=False
    )

    print("\nTOP 10 MOST REVIEWED PRODUCTS")
    display(product_summary.head(10))

    print("\nTOP 10 PRODUCTS BY AVERAGE RATING (MINIMUM 5 REVIEWS)")
    eligible_products = product_summary[
        product_summary["review_count"] >= 5
    ].sort_values("average_rating", ascending=False)

    display(eligible_products.head(10))

    product_summary.to_csv(
        os.path.join(OUTPUT_DIR, "product_level_summary.csv"),
        index=False
    )

# ============================================================
# 15. STATISTICAL TESTS
# ============================================================

statistical_results = []

# Spearman correlation: review length and rating
correlation_data = eda_df[
    ["review_length_words", "rating"]
].dropna()

if len(correlation_data) >= 3:
    corr, p_value = spearmanr(
        correlation_data["review_length_words"],
        correlation_data["rating"]
    )

    statistical_results.append({
        "test": "Spearman: review length vs rating",
        "statistic": corr,
        "p_value": p_value,
        "interpretation": (
            "Statistically significant association"
            if p_value < 0.05
            else "No statistically significant association detected"
        )
    })

# Chi-square: verified purchase and low rating
if "verified_purchase_clean" in eda_df.columns:
    test_df = eda_df.dropna(
        subset=["verified_purchase_clean", "low_rating"]
    )

    if (
        not test_df.empty
        and test_df["verified_purchase_clean"].nunique() == 2
        and test_df["low_rating"].nunique() == 2
    ):
        contingency = pd.crosstab(
            test_df["verified_purchase_clean"],
            test_df["low_rating"]
        )

        if contingency.shape == (2, 2):
            chi2, p_value, dof, expected = chi2_contingency(contingency)

            statistical_results.append({
                "test": "Chi-square: verified purchase vs low rating",
                "statistic": chi2,
                "p_value": p_value,
                "interpretation": (
                    "Evidence of association"
                    if p_value < 0.05
                    else "No statistically significant association detected"
                )
            })

# Mann-Whitney U: rating distributions for verified/unverified reviews
if "verified_purchase_clean" in eda_df.columns:
    verified_ratings = eda_df.loc[
        eda_df["verified_purchase_clean"] == 1, "rating"
    ].dropna()

    unverified_ratings = eda_df.loc[
        eda_df["verified_purchase_clean"] == 0, "rating"
    ].dropna()

    if len(verified_ratings) > 0 and len(unverified_ratings) > 0:
        statistic, p_value = mannwhitneyu(
            verified_ratings,
            unverified_ratings,
            alternative="two-sided"
        )

        statistical_results.append({
            "test": "Mann-Whitney U: verified vs unverified ratings",
            "statistic": statistic,
            "p_value": p_value,
            "interpretation": (
                "Evidence of a distribution difference"
                if p_value < 0.05
                else "No statistically significant difference detected"
            )
        })

stats_df = pd.DataFrame(statistical_results)

print("\nSTATISTICAL TEST RESULTS")
if not stats_df.empty:
    display(stats_df)
    stats_df.to_csv(
        os.path.join(OUTPUT_DIR, "statistical_test_results.csv"),
        index=False
    )
else:
    print("No statistical tests could be completed with available columns.")

# ============================================================
# 16. FEATURE DICTIONARY
# ============================================================

feature_descriptions = {
    "review_length_chars": "Number of characters in review text",
    "review_length_words": "Number of words in review text",
    "title_length_chars": "Number of characters in review title",
    "positive_review": "1 if rating is at least 4, otherwise 0",
    "low_rating": "1 if rating is 1 or 2, otherwise 0",
    "neutral_rating": "1 if rating is 3, otherwise 0",
    "has_helpful_vote": "1 if helpful vote count is greater than zero",
    "verified_purchase_clean": "Normalized verified purchase indicator",
    "analysis_date": "Review date parsed from timestamp or review_date",
    "review_year": "Year of review",
    "review_month": "Month of review",
    "review_quarter": "Quarter of review",
    "review_day_of_week": "Day of week, Monday=0 through Sunday=6"
}

feature_dictionary = pd.DataFrame(
    [
        {"feature": feature, "description": description}
        for feature, description in feature_descriptions.items()
        if feature in eda_df.columns
    ]
)

print("\nFEATURE DICTIONARY")
display(feature_dictionary)

feature_dictionary.to_csv(
    os.path.join(OUTPUT_DIR, "feature_dictionary.csv"),
    index=False
)

# ============================================================
# 17. SAVE ENGINEERED DATASET
# ============================================================

featured_path = os.path.join(
    OUTPUT_DIR, "beauty_reviews_week4_featured.csv"
)

eda_df.to_csv(featured_path, index=False)

print("\nFEATURED DATASET SHAPE:", eda_df.shape)
print("FEATURED DATASET SAVED:", featured_path)

# ============================================================
# 18. FINAL VALIDATION SUMMARY
# ============================================================

validation_rows = [
    ["Raw dataset rows", len(raw_df)],
    ["Cleaned dataset rows", len(clean_df)],
    ["Featured dataset rows", len(eda_df)],
    ["Raw dataset columns", len(raw_df.columns)],
    ["Cleaned dataset columns", len(clean_df.columns)],
    ["Featured dataset columns", len(eda_df.columns)],
    ["New engineered columns", len(new_features)],
    ["Missing cells in cleaned data", int(clean_df.isna().sum().sum())],
    ["Duplicate rows in cleaned data", int(clean_df.duplicated().sum())],
    ["Mean rating", round(clean_df["rating"].mean(), 4)],
    ["Median rating", round(clean_df["rating"].median(), 4)],
    ["Valid timestamps in featured data", int(eda_df["analysis_date"].notna().sum())]
]

validation_summary = pd.DataFrame(
    validation_rows, columns=["metric", "value"]
)

print("\nFINAL VALIDATION SUMMARY")
display(validation_summary)

validation_summary.to_csv(
    os.path.join(OUTPUT_DIR, "week4_validation_summary.csv"),
    index=False
)

# ============================================================
# 19. SAVE OUTPUTS AS ZIP
# ============================================================

zip_path = (
    "/content/drive/MyDrive/Colab Notebooks/Dataset/"
    "Week_04_EDA_Outputs.zip"
)

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zipf:
    for filename in os.listdir(OUTPUT_DIR):
        full_path = os.path.join(OUTPUT_DIR, filename)

        if os.path.isfile(full_path):
            zipf.write(full_path, arcname=filename)

print("\n" + "=" * 65)
print("WEEK 04 ANALYSIS COMPLETED")
print("Output folder:", OUTPUT_DIR)
print("ZIP file:", zip_path)
print("=" * 65)
```
