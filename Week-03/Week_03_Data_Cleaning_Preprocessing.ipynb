```python
from google.colab import drive
drive.mount('/content/drive')
file_path = "/content/drive/My Drive/Colab Notebooks/Dataset/reviews_sample.parquet"

df = pd.read_parquet(file_path)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 records:")
display(df.head())
```

```python
print("========== DATA QUALITY REPORT ==========")

print("\n1. Dataset Information")
print("Total rows:", len(df))
print("Total columns:", len(df.columns))

print("\n2. Missing Values")
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().mean() * 100).round(2)
})
display(missing_report)

print("\n3. Duplicate Analysis")
print("Exact duplicate rows:", df.duplicated().sum())

print("\n4. Rating Analysis")
print(df["rating"].value_counts(dropna=False).sort_index())
print("Invalid ratings:",
      (~df["rating"].between(1, 5)).sum())

print("\n5. Helpful Vote Analysis")
print(df["helpful_vote"].describe())
print("Negative helpful votes:",
      (df["helpful_vote"] < 0).sum())

print("\n6. Verified Purchase Values")
print(df["verified_purchase"].value_counts(dropna=False))

print("\n7. Timestamp Analysis")
print("Minimum timestamp:", df["timestamp"].min())
print("Maximum timestamp:", df["timestamp"].max())

print("\n8. Data Types")
print(df.dtypes)
```

```python
# Step 1: Check blank and whitespace-only text
print("Blank values in title:",
      df["title"].apply(lambda x: isinstance(x, str) and not x.strip()).sum())

print("Blank values in review text:",
      df["text"].apply(lambda x: isinstance(x, str) and not x.strip()).sum())

# Step 2: Display duplicate examples
print("\nDuplicate records:")
display(df[df.duplicated(keep=False)].head(10))

# Step 3: Inspect timestamps as readable dates
df["review_date"] = pd.to_datetime(
    df["timestamp"], unit="s", utc=True, errors="coerce"
)

print("\nReview date range:")
print("Earliest:", df["review_date"].min())
print("Latest:", df["review_date"].max())

print("\nInvalid timestamps:", df["review_date"].isna().sum())
```

```python
# Preserve original data and remove exact duplicate rows
df_clean = df.copy()

rows_before = len(df_clean)

df_clean = df_clean.drop_duplicates().copy()

rows_after = len(df_clean)
duplicates_removed = rows_before - rows_after

print("Rows before cleaning:", rows_before)
print("Duplicate rows removed:", duplicates_removed)
print("Rows after cleaning:", rows_after)
print("Remaining exact duplicates:", df_clean.duplicated().sum())
```

```python
import re
import html

def clean_review(value):
    if pd.isna(value):
        return ""

    value = str(value)
    value = html.unescape(value)
    value = re.sub(r"<[^>]+>", " ", value)
    value = re.sub(r"\s+", " ", value)
    value = value.strip()

    return value

df_clean["title_clean"] = df_clean["title"].apply(clean_review)
df_clean["text_clean"] = df_clean["text"].apply(clean_review)

print("Text cleaning completed!")

display(
    df_clean[["title", "title_clean", "text", "text_clean"]].head(5)
)
```

```python
df_clean["review_length"] = df_clean["text_clean"].str.len()

print("Review length statistics:")
display(df_clean["review_length"].describe())

print("\nShortest reviews:")
display(
    df_clean.nsmallest(5, "review_length")[
        ["rating", "title_clean", "text_clean", "review_length"]
    ]
)

print("\nLongest reviews:")
display(
    df_clean.nlargest(5, "review_length")[
        ["rating", "title_clean", "text_clean", "review_length"]
    ]
)
```

```python
df_clean["review_date"] = pd.to_datetime(
    df_clean["timestamp"],
    unit="s",
    utc=True,
    errors="coerce"
)

print("Invalid timestamps:", df_clean["review_date"].isna().sum())
print("Date range:")
print(df_clean["review_date"].min())
print(df_clean["review_date"].max())
```

```python
print("========== FINAL DATA VALIDATION ==========")

print("\n1. Dataset Shape")
print("Original dataset:", df.shape)
print("Cleaned dataset:", df_clean.shape)

print("\n2. Duplicate Check")
print("Remaining exact duplicates:", df_clean.duplicated().sum())

print("\n3. Missing Values")
display(df_clean.isnull().sum())

print("\n4. Rating Validation")
print("Minimum rating:", df_clean["rating"].min())
print("Maximum rating:", df_clean["rating"].max())
print("Invalid ratings:", (~df_clean["rating"].between(1, 5)).sum())

print("\n5. Text Validation")
print("Empty cleaned reviews:", (df_clean["text_clean"] == "").sum())

print("\n6. Timestamp Validation")
print("Invalid timestamps:", df_clean["review_date"].isna().sum())

print("\n7. Final Columns")
print(df_clean.columns.tolist())
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Exact Duplicate Rows",
        "Missing Values",
        "Invalid Ratings",
        "Empty Cleaned Reviews",
        "Invalid Timestamps"
    ],
    "Before Cleaning": [
        len(df),
        df.duplicated().sum(),
        int(df.isnull().sum().sum()),
        int((~df["rating"].between(1, 5)).sum()),
        int(df["text"].apply(
            lambda x: isinstance(x, str) and not x.strip()
        ).sum()),
        int(pd.to_datetime(
            df["timestamp"], unit="s", utc=True, errors="coerce"
        ).isna().sum())
    ],
    "After Cleaning": [
        len(df_clean),
        df_clean.duplicated().sum(),
        int(df_clean.isnull().sum().sum()),
        int((~df_clean["rating"].between(1, 5)).sum()),
        int((df_clean["text_clean"] == "").sum()),
        int(df_clean["review_date"].isna().sum())
    ]
})

display(summary)
```

```python
df_clean.to_csv(
    "/content/drive/My Drive/Colab Notebooks/Dataset/beauty_reviews_cleaned.csv",
    index=False
)

summary.to_csv(
    "/content/drive/My Drive/Colab Notebooks/Dataset/week3_cleaning_summary.csv",
    index=False
)

print("Files saved successfully!")
print("1. beauty_reviews_cleaned.csv")
print("2. week3_cleaning_summary.csv")
```
